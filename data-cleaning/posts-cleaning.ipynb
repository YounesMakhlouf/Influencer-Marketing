```python
import logging

from services.influencers_service import InfluencersService
from shared.mongo import MongoConnection
from web.cleaning import clean_and_process_content

logging.basicConfig(level=logging.INFO, format="%(asctime)s - %(levelname)s - %(message)s")
```

```python
def clean_and_process_posts(posts: list[dict], key: str) -> None:
    """Clean every post's *key* field using the shared cleaning pipeline."""
    for post in posts:
        content = post.get(key, "")
        if not content:
            continue
        cleaned = clean_and_process_content(content)
        try:
            influencers_service.update_post(post, key, cleaned)
        except Exception as e:
            logging.error(f"Error updating post: {e}.")
```

```python
mongo_connection = MongoConnection()
influencers_service = InfluencersService(mongo_connection)
posts = influencers_service.get_posts()

clean_and_process_posts(posts, "posts")
```
